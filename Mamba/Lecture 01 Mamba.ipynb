```python
from sklearn.datasets import fetch_california_housing
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split as split

import torch
from torch.utils.data import DataLoader, TensorDataset
```

```python
data = fetch_california_housing()
X_train, X_test, y_train, y_test = split(data.data, data.target, test_size=0.2, random_state=42)
```

# Scaling

```python
scaler = StandardScaler()
scaled_train_x = scaler.fit_transform(X_train)
scaled_test_x = scaler.transform(X_test)
```

# Conversion to Pytorch Tensor

```python
tensor_train_x = torch.tensor(scaled_train_x, dtype=torch.float32)
tensor_test_x = torch.tensor(scaled_test_x, dtype=torch.float32)
```

```python
tensor_train_y = torch.tensor(y_train, dtype=torch.float32).view(-1, 1)
tensor_test_y = torch.tensor(y_test, dtype=torch.float32).view(-1, 1)
```

```python
train_dataloader = DataLoader(TensorDataset(tensor_train_x, tensor_train_y), batch_size=32)
```
